```python
import seaborn as sns
import pandas as pd
import plotly.express as px
```

```python
# Load Dataset
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# Line Plot
```

```python
px.line(tips, y='total_bill', title='Total Bill per Order')
```

```python
# Scatter Plot
```

```python
px.scatter(tips, x='total_bill', y='tip', color='sex',
           size='size', hover_data=['day', 'time'],
           title='Total Bill vs Tip')
```

```python
# Bar Plot
```

```python
px.bar(tips, x='day', y='total_bill', color='sex',
       barmode='group', title='Total Bill by Day')
```

```python
# Histogram
```

```python
px.histogram(tips, x='total_bill', color='smoker',
             nbins=30, title='Distribution of Total Bill')
```

```python
# Box Plot
```

```python
px.box(tips, x='day', y='tip', color='smoker',
       points='all', title='Tip Distribution by Day')
```

```python
# Pie Chart
```

```python
px.pie(tips, names='day', values='total_bill',
       title='Total Bill Share by Day')
```
