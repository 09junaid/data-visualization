```python
import matplotlib.pyplot as plt
x=['Mon','Tue','Wed','Thur','Friday']  # x-axis
y=[10,15,7,20,12]                      # y-axis

plt.plot(x,y)

plt.title('Bakery Sales This Week!')
plt.xlabel('Days of the week')
plt.ylabel('Sales Per Day')
plt.show()
```
