```python
# subplot(nrows, ncols, index) index not 0 base it's 1 base
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,20,15,25]

plt.subplot(1,2,1)   # 1 row, 2nd col, 1st position num
plt.plot(x,y,color='blue')
plt.title('Line Chart')

plt.subplot(1,2,2)  # 1 row, 2nd col, 2nd positon num
plt.bar(x,y,color='green')
plt.title("Bar Chart")

plt.suptitle('Comparison of Line Chart and Bar Chart')

plt.tight_layout()
plt.show()
```
