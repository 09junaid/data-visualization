```python

# 1) Data Point
# -----------------------------------------------------------
# A data point is a single piece of information plotted on a graph.
# It is represented as a coordinate pair (x, y), where:
#   - x is the position along the horizontal axis
#   - y is the position along the vertical axis
#
# Example: if you are tracking temperature over days,
# (3, 36) means "on day 3, the temperature was 36°C".
# Every value in your dataset becomes one data point on the plot.
#
# In matplotlib, when you call plt.plot([1,2,3], [10,20,15]),
# you are passing 3 data points: (1,10), (2,20), (3,15).

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]          # x-coordinates of each data point
y = [10, 20, 15, 30, 25]     # y-coordinates of each data point

plt.plot(x, y)
plt.title("Data Points Demo")
plt.show()
```

```python

# 2) x-axis and y-axis
# -----------------------------------------------------------
# The x-axis is the HORIZONTAL line at the bottom of the graph.
# The y-axis is the VERTICAL line on the left side of the graph.
# Together they form a coordinate system (called a Cartesian plane).
#
# - x-axis usually represents the INDEPENDENT variable
#   (something you control or measure, like time, age, categories)
# - y-axis usually represents the DEPENDENT variable
#   (the outcome that changes because of x, like price, score, count)
#
# plt.xlabel() → adds a text label below the x-axis
# plt.ylabel() → adds a text label to the left of the y-axis

import matplotlib.pyplot as plt

days  = [1, 2, 3, 4, 5]      # independent variable → goes on x-axis
sales = [200, 350, 300, 450, 400]  # dependent variable → goes on y-axis

plt.plot(days, sales)
plt.xlabel("Day")             # label for the x-axis
plt.ylabel("Sales ($)")       # label for the y-axis
plt.title("x-axis and y-axis Demo")
plt.show()
```

```python

# 3) Figure
# -----------------------------------------------------------
# A Figure is the ENTIRE window or canvas on which everything is drawn.
# Think of it as a blank sheet of paper before you start drawing.
#
# Key properties of a Figure:
#   - figsize=(width, height) → size of the canvas in inches
#   - dpi           → resolution (dots per inch)
#   - facecolor     → background color of the canvas
#
# A Figure can contain ONE or MANY Axes (subplots).
# You create a Figure explicitly with:
#   fig = plt.figure()
# or implicitly when you call plt.plot() (matplotlib creates one for you).
#
# The Figure is the outermost container — everything else lives inside it.

import matplotlib.pyplot as plt

# Explicitly creating a Figure with a custom size
fig = plt.figure(figsize=(8, 4), facecolor="lightyellow")

ax = fig.add_subplot(1, 1, 1)   # add one Axes to the Figure
ax.plot([1, 2, 3], [4, 5, 6])
ax.set_title("Figure Demo")
plt.show()
```

```python

# 4) Axes / Graph-box
# -----------------------------------------------------------
# An Axes is the actual PLOTTING AREA inside a Figure — the box where
# your data is drawn with its own x-axis, y-axis, ticks, and labels.
#
# Difference between Figure and Axes:
#   Figure  = the whole canvas (the picture frame)
#   Axes    = the drawing area inside the frame (the canvas itself)
#
# One Figure can contain multiple Axes (subplots side by side).
# The Axes object (ax) gives you precise control:
#   ax.plot()       → draw data
#   ax.set_title()  → set the subplot title
#   ax.set_xlabel() → label the x-axis of this specific subplot
#   ax.set_xlim()   → control the visible x range
#   ax.set_ylim()   → control the visible y range
#
# plt.subplots() is the easiest way to get both Figure and Axes at once.

import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))  # 1 row, 2 columns → 2 Axes

axes[0].plot([1, 2, 3], [1, 4, 9])
axes[0].set_title("Left Axes")

axes[1].plot([1, 2, 3], [9, 4, 1])
axes[1].set_title("Right Axes")

plt.suptitle("Axes / Graph-box Demo")
plt.show()
```

```python

# 5) Marker
# -----------------------------------------------------------
# A marker is the SYMBOL drawn at each individual data point on the plot.
# By default, plt.plot() draws only a line with NO markers.
# Adding markers makes each data point visible as a distinct symbol.
#
# Common marker values:
#   'o'  → filled circle
#   's'  → square
#   '^'  → upward triangle
#   'D'  → diamond
#   'x'  → cross / X
#   '+'  → plus sign
#   '*'  → star
#
# Marker styling arguments:
#   marker      → shape of the marker (e.g., 'o', 's', '^')
#   markersize  → size of the marker in points
#   markerfacecolor → fill color of the marker
#   markeredgecolor → border color of the marker

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 4]

plt.plot(x, y,
         marker='o',            # circle at each data point
         markersize=10,         # size of the circle
         markerfacecolor='red', # inside of the circle = red
         markeredgecolor='black') # border of the circle = black
plt.title("Marker Demo")
plt.show()
```

```python

# 6) Plot
# -----------------------------------------------------------
# A plot is the VISUAL REPRESENTATION of data inside an Axes.
# "Plotting" means translating numbers into a picture so patterns,
# trends, and relationships become easy to see.
#
# matplotlib supports many plot types:
#   plt.plot()      → line plot (connects data points with a line)
#   plt.scatter()   → scatter plot (dots, no connecting line)
#   plt.bar()       → vertical bar chart
#   plt.barh()      → horizontal bar chart
#   plt.hist()      → histogram (frequency distribution)
#   plt.pie()       → pie chart
#   plt.boxplot()   → box-and-whisker plot
#
# The word "plot" is also used loosely to refer to the entire chart/graph.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 3, 6, 2, 8]

# Line plot
plt.subplot(1, 3, 1)
plt.plot(x, y)
plt.title("Line Plot")

# Scatter plot
plt.subplot(1, 3, 2)
plt.scatter(x, y)
plt.title("Scatter Plot")

# Bar plot
plt.subplot(1, 3, 3)
plt.bar(x, y)
plt.title("Bar Plot")

plt.tight_layout()
plt.show()
```

```python

# 7) Line Style
# -----------------------------------------------------------
# Line style controls how the LINE connecting data points LOOKS.
# It only applies to plots that draw lines (e.g., plt.plot()).
#
# Common linestyle values:
#   '-'   → solid line       (default)
#   '--'  → dashed line
#   '-.'  → dash-dot line
#   ':'   → dotted line
#   'None'→ no line at all (only markers are shown)
#
# You can also control:
#   linewidth (or lw) → thickness of the line in points
#
# Shortcut: combine color, marker, and linestyle in one string
#   'r--'  → red dashed line
#   'bo-'  → blue solid line with circle markers
#   'g:^'  → green dotted line with triangle markers

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

plt.plot(x, [5, 3, 6, 2, 8], linestyle='-',  linewidth=2, label="solid")
plt.plot(x, [4, 2, 5, 1, 7], linestyle='--', linewidth=2, label="dashed")
plt.plot(x, [3, 1, 4, 0, 6], linestyle='-.',  linewidth=2, label="dash-dot")
plt.plot(x, [2, 0, 3,-1, 5], linestyle=':',  linewidth=2, label="dotted")

plt.legend()
plt.title("Line Style Demo")
plt.show()
```

```python

# 8) Color
# -----------------------------------------------------------
# Color sets the VISUAL COLOR of plot elements (lines, markers, bars, text).
# matplotlib accepts colors in many formats:
#
#   Named colors  : 'red', 'blue', 'green', 'orange', 'purple', 'black'...
#   Short codes   : 'r'=red, 'g'=green, 'b'=blue, 'c'=cyan,
#                   'm'=magenta, 'y'=yellow, 'k'=black, 'w'=white
#   Hex codes     : '#FF5733'  (6-digit hexadecimal RGB)
#   RGB tuple     : (0.2, 0.4, 0.8)  → values between 0.0 and 1.0
#   RGBA tuple    : (0.2, 0.4, 0.8, 0.5) → 4th value is transparency (alpha)
#
# The `alpha` parameter (0.0 = fully transparent, 1.0 = fully opaque)
# controls how see-through a plot element is.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

plt.plot(x, [5,3,6,2,8], color='red',           label="named color")
plt.plot(x, [4,2,5,1,7], color='#2196F3',       label="hex color")
plt.plot(x, [3,1,4,0,6], color=(0.1,0.7,0.1),  label="RGB tuple")
plt.plot(x, [2,0,3,-1,5],color='purple', alpha=0.4, label="alpha (transparent)")

plt.legend()
plt.title("Color Demo")
plt.show()
```

```python

# 9) Legend
# -----------------------------------------------------------
# A legend is a SMALL BOX on the plot that explains what each line,
# bar, or scatter series represents — it maps colors/styles to labels.
#
# How it works:
#   1. Assign a label= to each plot call → e.g., plt.plot(..., label="Sales")
#   2. Call plt.legend() to make the box appear
#
# Positioning with loc=:
#   'best'         → matplotlib auto-picks the least-crowded spot (default)
#   'upper right'  → top-right corner
#   'upper left'   → top-left corner
#   'lower right'  → bottom-right corner
#   'center'       → center of the plot
#
# Extra options:
#   frameon=False  → remove the box border
#   fontsize=      → change legend text size
#   title=         → add a heading inside the legend box

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

plt.plot(x, [5, 3, 6, 2, 8], label="Revenue",  color='blue')
plt.plot(x, [3, 4, 2, 5, 3], label="Expenses", color='red')

plt.legend(loc='upper right', title="Financials", fontsize=10)
plt.title("Legend Demo")
plt.show()
```

```python

# 10) Label
# -----------------------------------------------------------
# A label is TEXT that describes an element of the plot.
# There are TWO main types of labels in matplotlib:
#
# Type 1 — Axis Labels:
#   Describe what the x-axis and y-axis represent.
#   plt.xlabel("Time (days)")   → text below the x-axis
#   plt.ylabel("Temperature °C") → text to the left of the y-axis
#
# Type 2 — Series Label (for the legend):
#   Assigned inside a plot call with the label= argument.
#   plt.plot(x, y, label="Temperature")
#   Only appears on the chart AFTER plt.legend() is called.
#
# Labels give context — without them, viewers cannot know what
# the axes or the colored lines actually mean.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [22, 25, 19, 28, 24]

plt.plot(x, y, label="Temperature", color='tomato', marker='o')

plt.xlabel("Day of Week")          # axis label for x
plt.ylabel("Temperature (°C)")     # axis label for y
plt.legend()                       # shows the series label "Temperature"
plt.title("Label Demo")
plt.show()
```

```python

# 11) Title
# -----------------------------------------------------------
# A title is the HEADING of the entire plot or a specific subplot.
# It sits at the TOP of the Axes and tells the viewer what the chart is about.
#
# Two kinds of title:
#   plt.title("My Chart")      → title for the current Axes (subplot title)
#   plt.suptitle("My Figure")  → super-title for the whole Figure
#                                 (sits above all subplots)
#
# Styling options:
#   fontsize  → size of the title text in points
#   color     → color of the title text
#   loc       → 'left', 'center' (default), or 'right' alignment
#   pad       → extra space between title and the plot in points
#
# A clear, descriptive title is one of the most important elements
# of a professional chart.

import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot([1,2,3], [4,5,6])
axes[0].set_title("Left Subplot Title", fontsize=12, color='navy', loc='left')

axes[1].plot([1,2,3], [6,5,4])
axes[1].set_title("Right Subplot Title", fontsize=12, color='darkgreen')

plt.suptitle("Figure-level Super Title", fontsize=16, color='maroon')
plt.tight_layout()
plt.show()
```

```python

# 12) Grid
# -----------------------------------------------------------
# A grid is a set of faint REFERENCE LINES drawn across the Axes background,
# extending from the tick marks on the x-axis and y-axis.
# Grids make it easier to read and estimate values at any data point.
#
# plt.grid(True)   → turn grid ON  (both x and y grid lines)
# plt.grid(False)  → turn grid OFF (default in most styles)
#
# You can control which grid lines appear:
#   axis='x'    → only vertical grid lines (from x-axis ticks)
#   axis='y'    → only horizontal grid lines (from y-axis ticks)
#   axis='both' → both (default)
#
# Styling options:
#   color     → color of the grid lines
#   linestyle → style of lines ('-', '--', ':', '-.')
#   linewidth → thickness of the grid lines
#   alpha     → transparency (0 = invisible, 1 = fully visible)

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [4, 2, 7, 3, 9]

plt.plot(x, y, marker='o')
plt.grid(True,
         axis='both',
         color='gray',
         linestyle='--',
         linewidth=0.7,
         alpha=0.6)
plt.title("Grid Demo")
plt.show()
```

```python

# 13) Function
# -----------------------------------------------------------
# A function is a REUSABLE BLOCK OF CODE that performs a specific task.
# You define it once and call it as many times as needed.
#
# Syntax:
#   def function_name(parameters):
#       """Optional docstring."""
#       # code block
#       return result
#
# In matplotlib, everything you type like plt.plot(), plt.show(),
# plt.title() are CALLS to built-in functions inside the pyplot module.
#
# Why functions matter for data visualization:
#   - Avoid repeating the same plotting code for multiple datasets
#   - Keep your notebook clean and organized
#   - Change the chart style in one place, and it updates everywhere
#
# A function that returns nothing still executes its body (e.g., plt.show()).

import matplotlib.pyplot as plt

# Define a reusable plotting function
def plot_data(x, y, title="My Chart", color="blue"):
    """Plot x vs y with a title and color."""
    plt.plot(x, y, color=color, marker='o')
    plt.title(title)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.show()

# Call the function — reuse it with different data
plot_data([1,2,3,4,5], [2,4,6,8,10], title="Linear Growth",  color="green")
plot_data([1,2,3,4,5], [1,4,9,16,25], title="Quadratic Growth", color="red")
```

```python

# 14) Method
# -----------------------------------------------------------
# A method is a FUNCTION that BELONGS TO AN OBJECT (a class instance).
# It is called using dot notation: object.method()
#
# Difference between a Function and a Method:
#   Function → standalone, called directly:  plt.plot()
#   Method   → tied to an object, called on it: ax.plot(), fig.savefig()
#
# In matplotlib's Object-Oriented API, you work with objects:
#   fig  (Figure object)  → has methods like fig.savefig(), fig.set_size_inches()
#   ax   (Axes object)    → has methods like ax.plot(), ax.set_title(),
#                           ax.set_xlabel(), ax.legend(), ax.grid()
#
# The Axes object (ax) is the most important — almost everything you do
# to customize a subplot is a method call on ax.

import matplotlib.pyplot as plt

fig, ax = plt.subplots()    # fig = Figure object, ax = Axes object

# These are all METHOD CALLS on the ax object:
ax.plot([1, 2, 3], [4, 5, 6], color='purple', marker='s')  # draw data
ax.set_title("Method Demo")                                  # set title
ax.set_xlabel("X-axis")                                      # label x
ax.set_ylabel("Y-axis")                                      # label y
ax.grid(True)                                                # enable grid

# Method call on the fig object:
fig.set_facecolor("lightyellow")    # change figure background color

plt.show()
```

```python

# 15) Parameters
# -----------------------------------------------------------
# Parameters are the VARIABLES listed inside a function's definition (def line).
# They act as placeholders for values that will be passed in when the function
# is called. Each call can pass different values, making the function flexible.
#
# Two types of parameters:
#   Positional parameters → must be passed in the correct ORDER
#   Default parameters    → have a fallback value, so they are optional
#
# In matplotlib:
#   plt.plot(x, y, color='blue', linewidth=2)
#      x, y      → positional parameters (required, order matters)
#      color     → keyword parameter with a default (optional)
#      linewidth → keyword parameter with a default (optional)
#
# Understanding parameters lets you READ the matplotlib documentation
# and know exactly which arguments a function accepts and which are optional.

import matplotlib.pyplot as plt

# Custom function showing positional vs default parameters
def draw_line(x, y, color='blue', linewidth=1, marker='o'):
    # x, y       → positional (required)
    # color      → default = 'blue'  (optional)
    # linewidth  → default = 1       (optional)
    # marker     → default = 'o'     (optional)
    plt.plot(x, y, color=color, linewidth=linewidth, marker=marker)

x = [1, 2, 3, 4, 5]

draw_line(x, [1,4,9,16,25])                         # uses all defaults
draw_line(x, [2,3,5,7,11], color='red', linewidth=3) # overrides color & width

plt.title("Parameters Demo")
plt.show()
```

```python

# 16) Keyword Arguments (kwargs)
# -----------------------------------------------------------
# A keyword argument is an argument passed to a function using the
# NAME=VALUE syntax, so the ORDER does not matter.
#
#   Positional: plt.plot([1,2,3], [4,5,6])        → order matters
#   Keyword:    plt.plot(x=[1,2,3], y=[4,5,6])    → order does NOT matter
#
# In matplotlib, most styling options are keyword arguments:
#   color=, linewidth=, linestyle=, marker=, label=, alpha=, etc.
#
# **kwargs (double-star kwargs):
#   Some functions accept **kwargs, meaning they accept ANY number of
#   keyword arguments and pass them along to another function internally.
#   This is how matplotlib's plt.plot() forwards styling to the Line2D object.
#
# Best practice: always use keyword arguments for optional styling
# so your code is self-documenting and readable.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 6, 2, 8, 4]

# All styling options below are KEYWORD ARGUMENTS
plt.plot(
    x, y,
    color='darkorange',    # kwarg: line color
    linewidth=2.5,         # kwarg: line thickness
    linestyle='--',        # kwarg: line style
    marker='D',            # kwarg: diamond marker
    markersize=8,          # kwarg: marker size
    alpha=0.85,            # kwarg: transparency
    label='My Data'        # kwarg: legend label
)
plt.legend()
plt.title("Keyword Arguments (kwargs) Demo")
plt.show()
```

```python

# 17) Object-Oriented API (OO API)
# -----------------------------------------------------------
# matplotlib has TWO interfaces for creating plots:
#
# Interface 1 — pyplot (state-machine) API  ← implicit, beginner-friendly
#   Uses plt.* functions. matplotlib keeps track of the "current" Figure
#   and Axes internally. You don't manage objects yourself.
#   e.g.: plt.plot(), plt.title(), plt.show()
#
# Interface 2 — Object-Oriented (OO) API  ← explicit, professional
#   You create and hold references to Figure and Axes objects,
#   then call methods directly on those objects.
#   e.g.: fig, ax = plt.subplots()  →  ax.plot()  →  ax.set_title()
#
# Why prefer the OO API?
#   - Full control over multiple subplots (you choose WHICH ax to modify)
#   - Cleaner, more readable code in scripts and applications
#   - Required for complex layouts, animations, and embedding in GUIs
#   - Industry standard in data science and production code
#
# Rule of thumb:
#   Quick one-off plot in a notebook → pyplot API is fine
#   Any serious or multi-subplot work → use the OO API

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 100)

# --- pyplot (state-machine) style ---
plt.figure(figsize=(5, 3))
plt.plot(x, np.sin(x), color='blue')
plt.title("pyplot API")
plt.show()

# --- Object-Oriented style ---
fig, ax = plt.subplots(figsize=(5, 3))  # explicit Figure and Axes objects
ax.plot(x, np.sin(x), color='green')   # method on the ax object
ax.set_title("Object-Oriented API")    # method on the ax object
plt.show()
```

```python

# 18) DPI (Dots Per Inch)
# -----------------------------------------------------------
# DPI stands for "Dots Per Inch". It measures the RESOLUTION of a figure —
# how many pixels are packed into every inch of the output image.
#
# How DPI works with figsize:
#   pixel width  = figsize_width_inches  × dpi
#   pixel height = figsize_height_inches × dpi
#
#   Example: figsize=(6, 4), dpi=100  →  600 × 400 pixels
#            figsize=(6, 4), dpi=200  →  1200 × 800 pixels (sharper image)
#
# When does DPI matter?
#   - Saving to PNG/JPG for reports: use dpi=150–300 for print quality
#   - Displaying in a Jupyter notebook: the default (~100) is usually fine
#   - Retina/HiDPI screens: higher dpi makes figures look crisp
#
# Common DPI values:
#   72   → screen-only (web)
#   100  → default in matplotlib
#   150  → good quality for presentations
#   300  → professional print quality

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 8, 6]

# Low resolution figure (72 dpi)
fig_low = plt.figure(figsize=(4, 3), dpi=72)
plt.plot(x, y, color='teal')
plt.title("Low DPI (72) — screen quality")
plt.show()

# High resolution figure (300 dpi) — ideal for saving to file
fig_high = plt.figure(figsize=(4, 3), dpi=300)
plt.plot(x, y, color='crimson')
plt.title("High DPI (300) — print quality")
# plt.savefig("high_res_plot.png", dpi=300)   # uncomment to save
plt.show()
```

```python

# 19) Backend
# -----------------------------------------------------------
# A backend is the ENGINE that matplotlib uses to RENDER (draw) figures
# and DISPLAY or SAVE them. The backend sits between your Python code
# and the screen/file output.
#
# Two categories of backends:
#
#   Interactive backends  → display a live window you can zoom/pan
#     'TkAgg'   → uses the Tkinter GUI toolkit (very common default)
#     'Qt5Agg'  → uses the Qt5 GUI toolkit
#     'WXAgg'   → uses the wxWidgets GUI toolkit
#     'MacOSX'  → native macOS rendering
#
#   Non-interactive backends  → render to file without showing a window
#     'Agg'     → Anti-Grain Geometry; renders to PNG in memory (default
#                 in scripts and Jupyter when no display is attached)
#     'SVG'     → renders to SVG vector format
#     'PDF'     → renders to PDF vector format
#     'PS'      → renders to PostScript format
#
# In Jupyter notebooks, a special "inline" backend is set by the magic:
#   %matplotlib inline   → renders PNGs embedded directly in the notebook
#   %matplotlib widget   → renders interactive figures in the notebook
#
# You rarely need to change the backend manually, but knowing it exists
# explains WHY plt.show() behaves differently in a script vs a notebook.

import matplotlib
import matplotlib.pyplot as plt

# Check which backend is currently active
print("Current backend:", matplotlib.get_backend())

# Switch backend programmatically (must be done BEFORE importing pyplot)
# matplotlib.use('Agg')   # uncomment to switch to the non-interactive Agg backend

x = [1, 2, 3, 4, 5]
y = [1, 3, 2, 5, 4]

plt.plot(x, y, color='slateblue', marker='o')
plt.title("Backend Demo")
plt.show()
```
