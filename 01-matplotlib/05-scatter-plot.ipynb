```python
# Scatter Plot
# plt.scatter(x,y,color='color_name',marker='marker_style',label='label_name')

import matplotlib.pyplot as plt

hours_studies = [1,2,3,4,5,6,7,8]
exam_scores = [50,55,60,65,70,75,80,85]

plt.scatter(hours_studies,exam_scores,color='red',marker='o',label='Student Data')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title('Relationship Between Study Time and Exam Score')
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter([1,2,3],[50,55,60],color='blue',label='Class A')  #g1
plt.scatter([1,2,3],[45,50,52],color='orange',label='Class B')  #g2

plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title('Comparison of two classes')
plt.legend()
plt.grid(True)

plt.show()
```
